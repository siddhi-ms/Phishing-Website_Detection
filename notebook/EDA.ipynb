# Phishing Website Detection - EDA
Run all cells from the `notebook/` folder. Plots are saved to `/assets`.

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
from src.preprocess import load_data, check_nulls, prepare_dataframe, save_eda_plots, FEATURES, TARGET, DATA_PATH
```

## 1. Load data (CSV = tabular storage)

```python
df = load_data(DATA_PATH)
print(df.shape)
df.head()
```

## 2. Null check and feature selection

```python
check_nulls(df)
data = prepare_dataframe(df)   # 12 features + CLASS_LABEL (flipped: 0 = Phishing, 1 = Legitimate)
data.head()
```

## 3. Summary statistics

```python
data.describe().T
```

## 4. Class distribution

```python
print(data[TARGET].value_counts())
data[TARGET].value_counts().plot(kind='pie', labels=['Legitimate' if i==1 else 'Phishing' for i in data[TARGET].value_counts().index], autopct='%1.1f%%')
plt.ylabel(''); plt.show()
```

## 5. Correlation heatmap

```python
import seaborn as sns
plt.figure(figsize=(10, 8))
sns.heatmap(data.corr(), annot=True, fmt='.2f', cmap='coolwarm', annot_kws={'size': 7})
plt.show()
```

## 6. Feature distributions by class

```python
fig, axes = plt.subplots(4, 3, figsize=(14, 14))
for ax, col in zip(axes.ravel(), FEATURES):
    sns.histplot(data=data, x=col, hue=TARGET, bins=25, ax=ax)
    ax.set_title(col)
plt.tight_layout(); plt.show()
```

## 7. Save all EDA images to /assets

```python
save_eda_plots(data)
```
